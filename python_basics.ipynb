## 4. Conditionals and even numbers

Finish `is_even` so that it returns `True` for even integers and `False` for odd integers. The `%` operator gives the remainder after division.

```python
# Your code
def is_even(number):
    # Return True when number is even, otherwise return False.
    return number % 2 == 0
```

```python
# Automatic check
assert is_even(4) is True, "4 is even."
assert is_even(7) is False, "7 is odd."
assert is_even(0) is True, "0 is even."
print("Exercise 4 passed!")
```

## 5. Lists and built-in functions

Make a list containing exactly five numbers, then use Python's `sum()` function to calculate their total.

```python
# Your code
numbers = [2, 4, 6, 8, 10]
sum_numbers = sum(numbers)  # Calculate the sum of the list.
```

```python
# Automatic check
assert isinstance(numbers, list) and len(numbers) == 5, "numbers must be a list of exactly five items."
assert all(isinstance(item, (int, float)) for item in numbers), "Each item in numbers should be numeric."
assert sum_numbers == sum(numbers), "sum_numbers should equal the sum of numbers."
print("Exercise 5 passed!")
```

## 6. Loops

Use a `for` loop to calculate the sum of the numbers from 1 through 5, inclusive, and store it in `loop_total`. Hint: `range(1, 6)` produces 1, 2, 3, 4, 5.

```python
# Your code
loop_total = 0
for number in range(1, 6):
    loop_total += number
```

```python
# Automatic check
assert loop_total == 15, "The sum of 1 + 2 + 3 + 4 + 5 should be 15."
print("Exercise 6 passed!")
```
